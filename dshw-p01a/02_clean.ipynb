```python
import glob
import os
import pandas as pd

project_dir = os.path.abspath("")
clean_dir = os.path.join(project_dir, "data", "clean")
combined_path = os.path.join(project_dir, "data", "combined", "combined_stocks.csv")

clean_files = sorted(glob.glob(os.path.join(clean_dir, "stock_*.csv")))
print(f"Clean stock files: {len(clean_files)}")
for p in clean_files[:5]:
    print("  ", os.path.basename(p), pd.read_csv(p, nrows=3).shape)

if os.path.exists(combined_path):
    combined = pd.read_csv(combined_path, parse_dates=["date"], dtype={"ticker": str})
    print("Combined data shape:", combined.shape)
    display(combined.head(5))
    print("Ticker summary:")
    print(combined.groupby("ticker")["close"].agg(["count", "mean", "std", "min", "max"]))
```
