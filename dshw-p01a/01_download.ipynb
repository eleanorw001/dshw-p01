```python
import glob
import os
import pandas as pd

base_dir = os.path.abspath("")
data_dirs = ["data/stock", "data/index", "data/macro", "data/finance"]
for d in data_dirs:
    files = sorted(glob.glob(os.path.join(base_dir, d, "*.csv")))
    print(f"{d} files: {len(files)}")
    for f in files[:3]:
        try:
            df = pd.read_csv(f, nrows=3)
            print(f"  {os.path.basename(f)} {df.shape}")
        except Exception as e:
            print(f"  {os.path.basename(f)} read error: {e}")

log_path = os.path.join(base_dir, "download_log.txt")
if os.path.exists(log_path):
    print("download_log.txt latest lines:")
    with open(log_path, encoding="utf-8") as f:
        print("".join(f.readlines()[-10:]))
```
