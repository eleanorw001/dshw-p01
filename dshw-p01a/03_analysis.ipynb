```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import analysis

sns.set_style("whitegrid")
print("Running analysis pipeline...")
capm_results, stats_summary, market_returns, macro_results = analysis.run_analysis()
print("Analysis pipeline completed.")
```

```python
output_dir = os.path.join(os.path.abspath(""), "output")
for name in ["capm_results.csv", "stats_summary.csv", "market_returns.csv", "macro_regression_results.csv"]:
    path = os.path.join(output_dir, name)
    print(name, "exists:", os.path.exists(path))
    if os.path.exists(path):
        df = pd.read_csv(path)
        print(df.shape)
        display(df.head(5))
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
market_returns = pd.read_csv(os.path.join(output_dir, "market_returns.csv"), parse_dates=["date"])
ax.plot(market_returns["date"], market_returns["mkt_ret"], color="tab:blue")
ax.set_title("市场日收益率（HS300）")
ax.set_xlabel("Date")
ax.set_ylabel("Daily Return")
plt.tight_layout()
plt.show()
```

```python
capm = pd.read_csv(os.path.join(output_dir, "capm_results.csv"))
capm_sorted = capm.sort_values("beta")
fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(x="beta", y="ticker", data=capm_sorted, palette="viridis", ax=ax)
ax.set_title("CAPM Beta 系数")
ax.set_xlabel("Beta")
ax.set_ylabel("Ticker")
plt.tight_layout()
plt.show()
```

```python
macro = pd.read_csv(os.path.join(output_dir, "macro_regression_results.csv"))
macro_sorted = macro.sort_values("beta_cpi")
fig, axes = plt.subplots(1, 2, figsize=(14, 5), constrained_layout=True)
sns.barplot(x="beta_cpi", y="ticker", data=macro_sorted, ax=axes[0], palette="coolwarm")
axes[0].set_title("CPI 对月度收益的回归系数")
sns.barplot(x="beta_m2", y="ticker", data=macro_sorted, ax=axes[1], palette="coolwarm")
axes[1].set_title("M2 对月度收益的回归系数")
plt.show()
```
